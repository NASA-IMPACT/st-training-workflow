```python
import os
import boto3

def download_prefix(bucket: str, prefix: str, local_dir: str):
    s3 = boto3.client('s3')
    paginator = s3.get_paginator("list_objects_v2")

    for page in paginator.paginate(Bucket=bucket, Prefix=prefix):
        for obj in page.get("Contents", []):
            key = obj["Key"]
            # preserve sub‑directory structure under local_dir
            rel_path = os.path.relpath(key, prefix)
            dest_path = os.path.join(local_dir, rel_path)
            os.makedirs(os.path.dirname(dest_path), exist_ok=True)
            s3.download_file(bucket, key, dest_path)
            print(f"↓ {key} → {dest_path}")


BUCKET = "nasa-smd-llm-data"
PREFIX = "nasa_eb_02/"
LOCAL_DIR = "./downloads/nasa-smd-llm-data/nasa_eb_02/"

download_prefix(BUCKET, PREFIX, LOCAL_DIR)
```
