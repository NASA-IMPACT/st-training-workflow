```python
import pandas as pd
import requests
import xmltodict
import logging
from tqdm.auto import tqdm
from typing import List, Dict, Any, Optional
from concurrent.futures import ThreadPoolExecutor, as_completed
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

# --- Setup Logging ---
# Configure logging to provide informative output
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

# --- Helper Functions ---

def deep_get(d: Dict, keys: List[str]) -> Optional[Any]:
    """
    Safely retrieves a nested value from a dictionary.

    Args:
        d: The dictionary to search.
        keys: A list of keys representing the path to the value.

    Returns:
        The requested value, or None if any key in the path is not found.
    """
    for key in keys:
        if isinstance(d, dict):
            d = d.get(key)
        else:
            return None
    return d

def get_session_with_retries(
    retries: int = 3,
    backoff_factor: float = 0.3,
    status_forcelist: tuple = (500, 502, 503, 504),
    session: Optional[requests.Session] = None,
) -> requests.Session:
    """
    Creates a requests.Session object with a retry mechanism.

    This setup will automatically retry requests that fail due to network errors
    or specific HTTP status codes, with an exponential backoff delay.
    """
    session = session or requests.Session()
    retry_strategy = Retry(
        total=retries,
        read=retries,
        connect=retries,
        backoff_factor=backoff_factor,
        status_forcelist=status_forcelist,
    )
    adapter = HTTPAdapter(max_retries=retry_strategy)
    session.mount('http://', adapter)
    session.mount('https://', adapter)
    return session

# --- Core Logic ---

def get_purpose_from_cmr_xml(url: str, session: requests.Session) -> Optional[List[str]]:
    """
    Fetches a CMR XML from a URL, parses it, and extracts the 'purpose' field(s).

    Args:
        url: The URL of the XML metadata to fetch.
        session: A requests.Session object to use for the HTTP request.

    Returns:
        A list of purpose strings found in the XML, or None if an error occurs.
    """
    try:
        response = session.get(url, timeout=15)  # Set a reasonable timeout
        response.raise_for_status()  # Raise an exception for bad status codes (4xx or 5xx)
        
        xml_content = response.content
        if not xml_content:
            logging.warning(f"Empty content received from URL: {url}")
            return [] # Return empty list for empty content

        xml_dict = xmltodict.parse(xml_content)

        # Safely navigate the nested dictionary structure
        id_info_path = ["gmi:MI_Metadata", "gmd:identificationInfo"]
        identification_info = deep_get(xml_dict, id_info_path)

        if not identification_info:
            logging.warning(f"Could not find 'identificationInfo' in XML from: {url}")
            return []

        # Ensure identification_info is a list to handle single or multiple entries
        if not isinstance(identification_info, list):
            identification_info = [identification_info]

        purposes = []
        purpose_path = ["gmd:MD_DataIdentification", "gmd:purpose", "gco:CharacterString"]
        for item in identification_info:
            purpose = deep_get(item, purpose_path)
            if purpose and isinstance(purpose, str):
                purposes.append(purpose.strip())
        
        return purposes

    except requests.exceptions.RequestException as e:
        logging.error(f"Request failed for URL {url}: {e}")
        return None
    except xmltodict.expat.ExpatError as e:
        logging.error(f"XML parsing failed for URL {url}: {e}")
        return None
    except Exception as e:
        logging.error(f"An unexpected error occurred for URL {url}: {e}")
        return None

def process_urls_concurrently(urls: pd.Series) -> pd.Series:
    """
    Processes a series of URLs concurrently to fetch and parse purposes.

    Args:
        urls: A pandas Series containing the URLs to process.

    Returns:
        A pandas Series containing the lists of purposes for each URL.
    """
    results = [None] * len(urls)
    session = get_session_with_retries()
    
    # Use ThreadPoolExecutor for concurrent I/O-bound tasks
    with ThreadPoolExecutor(max_workers=10) as executor:
        # Create a dictionary to map future objects to their original index
        future_to_index = {executor.submit(get_purpose_from_cmr_xml, url, session): i for i, url in enumerate(urls)}
        
        # Use tqdm to create a progress bar
        pbar = tqdm(total=len(urls), desc="Fetching CMR Purposes")
        
        for future in as_completed(future_to_index):
            index = future_to_index[future]
            try:
                results[index] = future.result()
            except Exception as e:
                # This catches errors from the future.result() call itself
                logging.error(f"Future for URL {urls[index]} generated an exception: {e}")
            pbar.update(1) # Increment the progress bar
        pbar.close()

    return pd.Series(results, index=urls.index)

df = pd.read_csv("/rhome/sawale/indus_traning/sentense_transformers/data/api_data/processed_dev_sde_index_api_dump_2025-06-19.csv")
cmr = df[df["treepath"] == "/Earth Science/Earth Science Data (CMR)/"]


cmr['purposes'] = process_urls_concurrently(cmr['url1'])
cols_to_convert = ['sourcestr25', 'sourcestr28', 'sourcevarchar7', 'sourcestr32', 'sourcestr34']
cmr[cols_to_convert] = cmr[cols_to_convert].astype(str)

cmr.to_parquet("/rhome/sawale/indus_traning/sentense_transformers/data/api_data/cmr_with_purposes.parquet", index=False)
```

```python
import pandas as pd
import numpy as np
import json

def transform_cmr_dataframe(cmr: pd.DataFrame) -> pd.DataFrame:
    """
    Transforms a raw CMR DataFrame into a structured format.

    This updated function vectorizes the creation of the 'context' column for
    improved performance and ensures all specified fields from the original
    request are consistently included in the 'metadata' column.

    Args:
        cmr: The input pandas DataFrame with the source CMR data.

    Returns:
        A new pandas DataFrame with the columns: 'query', 'context',
        'type', 'synthesized', 'source', and 'metadata'.
    """
    # Create a copy to avoid modifying the original DataFrame
    df = cmr.copy()

    # 1. Create the 'query' column from the 'title'
    df['query'] = df['title']

    # 2. Create the 'context' column using a more efficient, vectorized approach
    context_parts = []
    
    # Define the fields and their labels for the context string
    context_fields = {
        'data_product_desc': 'Description',
        'purposes': 'Purpose',
        'scientific_keywords_1': 'Scientific Keywords',
        'cmr_phenomena': 'CMR Phenomena',
        'dataformat': 'Dataformat',
        'instrument': 'Instrument',
        # 'platform': 'Platform',
        # 'agency': 'Agency',
        
    }
    
    for col, label in context_fields.items():
        # Ensure the column exists in the DataFrame
        if col in df.columns:
            # Use np.where for conditional, vectorized string creation
            part = np.where(
                df[col].notna() & (df[col] != ''),
                f"{label}: " + df[col].astype(str),
                '' # Use an empty string if the condition is false
            )
            context_parts.append(pd.Series(part, index=df.index))

    # Join all parts. Concatenate the series and then join non-empty parts with a newline.
    # This avoids leading/trailing/double newlines.
    df['context'] = [
        "\n".join(filter(None, values)) for values in zip(*context_parts)
    ]
    
    # 3. Create the 'synthesized' column
    df['synthesized'] = False
    
    # 4. Create 'type' and 'source' columns with default values
    df['type'] = 'title-description'
    df['source'] = 'CMR'

    # 5. Create the 'metadata' column
    # Define all columns to be included in the metadata JSON for consistency.
    metadata_cols = [
        'id', 'treepath', 'download_url', 'title', 'agency',
        'data_product_desc', 'scientific_keywords_1', 'platform',
        'instrument', 'cmr_phenomena', 'dataformat', 'purposes'
    ]
    
    # Filter out columns that don't exist in the input DataFrame to prevent errors.
    existing_metadata_cols = [col for col in metadata_cols if col in df.columns]

    # Efficiently create a list of dictionaries from the relevant columns.
    metadata_dicts = df[existing_metadata_cols].to_dict(orient='records')
    
    # Convert the list of dictionaries to a Series of JSON strings.
    df['metadata'] = pd.Series(
        [json.dumps(d) for d in metadata_dicts],
        index=df.index
    )

    # 6. Select and order the final columns
    final_columns = ['query', 'context', 'type', 'synthesized', 'source', 'metadata']
    final_df = df[final_columns]
    
    return final_df

cmr_pairs = transform_cmr_dataframe(cmr)
```

```python
import pandas as pd
import numpy as np
import json


def transform_pds_data(df: pd.DataFrame) -> pd.DataFrame:
    """
    Transforms a PDS DataFrame to a specified format with columns:
    query, context, type, synthesized, source, and metadata,
    based on the latest guidelines.

    Args:
        df: The input DataFrame with the source PDS columns.

    Returns:
        A new DataFrame with the transformed data.
    """
    # Create a new DataFrame to hold the transformed data
    final_df = pd.DataFrame()

    # 1. 'query' transformation
    # The guideline specifies 'title' as the source for query,
    # but the current columns list and previous code used 'data_item_title'.
    # Assuming 'title' is the correct new column name to use for 'query'.
    final_df['query'] = df['title']

    # 2. Create the 'context' column using a more efficient, vectorized approach
    context_parts = []
    
    # Define the fields and their labels for the context string
    context_fields = {
        'data_product_desc': 'Description',
        'scientific_keywords_1': 'Target',
        'repository': 'Repository',
        'investigation': 'Investigation',
        'instrument': 'Instrument',
    }

    for col, label in context_fields.items():
        # Ensure the column exists in the DataFrame
        if col in df.columns:
            # Use np.where for conditional, vectorized string creation
            part = np.where(
                df[col].notna() & (df[col] != ''),
                f"{label}: " + df[col].astype(str),
                '' # Use an empty string if the condition is false
            )
            context_parts.append(pd.Series(part, index=df.index))

    # This avoids leading/trailing/double newlines.
    final_df['context'] = [
        "\n".join(filter(None, values)) for values in zip(*context_parts)
    ]

    # 3. 'synthesized' transformation
    final_df['synthesized'] = False
    
    # 4. 'source' transformation
    final_df['source'] = 'PDS'

    # 5. 'type' column (not explicitly defined in new guidelines, keeping a default)
    final_df['type'] = 'title-description' # Assuming a more specific type than 'default'

    # 6. 'metadata' transformation
    metadata_cols = [
        'id', 'pds_lid', 'treepath', 'title', 'data_product_desc',
        'scientific_keywords_1', 'repository', 'instrument',
        'platform', 'investigation'
    ]
    # Filter out columns that don't exist in the input DataFrame to prevent errors.
    existing_metadata_cols = [col for col in metadata_cols if col in df.columns]
    # Efficiently create a list of dictionaries from the relevant columns.
    metadata_dicts = df[existing_metadata_cols].to_dict(orient='records')
    # Convert the list of dictionaries to a Series of JSON strings.
    final_df['metadata'] = pd.Series(
        [json.dumps(d) for d in metadata_dicts],
        index=df.index
    )

    # 6. Select and order the final columns
    final_columns = ['query', 'context', 'type', 'synthesized', 'source', 'metadata']
    final_df = final_df[final_columns]

    return final_df


pds_pairs = transform_pds_data(pds)
```
