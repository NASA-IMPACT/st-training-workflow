# analysis of text quality


```python
# plot distribution of length of text column in new_df

import matplotlib.pyplot as plt
plt.hist(new_df["text"].str.len(), bins=100)
plt.xlabel("Length of text")
plt.ylabel("Count")
plt.title("Distribution of text length in new SDE data")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.hist(new_df["text"].str.len(), bins=100000)
plt.xlim(0, 1000)  # Set x-axis limits from 0 to 10,000 characters
plt.xlabel("Length of text")
plt.ylabel("Count")
plt.title("Distribution of text length in new SDE data")
plt.show()
```

```python
# Calculate 5-point summary of text length
text_lengths = new_df["text"].str.len()
summary = text_lengths.describe()

print("5-Point Summary of Text Length:")
print(f"Minimum: {summary['min']:.0f}")
print(f"Q1 (25th percentile): {summary['25%']:.0f}")
print(f"Median (50th percentile): {summary['50%']:.0f}")
print(f"Q3 (75th percentile): {summary['75%']:.0f}")
print(f"Maximum: {summary['max']:.0f}")
print(f"Mean: {summary['mean']:.0f}")
print(f"Standard Deviation: {summary['std']:.0f}")
```

```python
# Using seaborn for better styling
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.boxplot(y=text_lengths)
plt.ylabel("Text Length (characters)")
plt.title("Box Plot of Text Length Distribution in New SDE Data")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# get text 20 samples of highest prob
top_samples = new_with_prob.sort_values(by="prob", ascending=False).head(20)
top_samples["text"].to_list()
for text in top_samples["text"].to_list()[:10]:
    print(text)
    print("-----")
```
