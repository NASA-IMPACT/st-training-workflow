```python
from datasets import load_dataset
from huggingface_hub import hf_hub_download
import json
import pandas as pd

x0 = pd.read_json("/rhome/sawale/indus_traning/sentense_transformers/data/NASA_Embedding_Traindata/title_abstract_prefix17.jsonl.gz", lines=True, compression='gzip')
x1 = pd.read_json("/rhome/sawale/indus_traning/sentense_transformers/data/NASA_Embedding_Traindata/title_abstract_prefix18.jsonl.gz", lines=True, compression='gzip')
x2 = pd.read_json("/rhome/sawale/indus_traning/sentense_transformers/data/NASA_Embedding_Traindata/title_abstract_prefix19.jsonl.gz", lines=True, compression='gzip')
x3 = pd.read_json("/rhome/sawale/indus_traning/sentense_transformers/data/NASA_Embedding_Traindata/title_abstract_prefix20.jsonl.gz", lines=True, compression='gzip')


df = pd.concat([x0, x1, x2, x3], ignore_index=True)

df.head()
```

```python
df.shape
```

```python
df["negatives"].iloc[0]
```

```python
# check if htere are multiple positives for a query

multiple_positives = 0
more_than_one_negative = 0

for i in (range(df.shape[0])):
    positives = df["positives"].iloc[i]["docs"]
    negatives = df["negatives"].iloc[i]["docs"]
    query = df["query"].iloc[i]

    # print(query)
    # print(positives)
    # print(negatives)
    # print("===" * 20)
    
    if len(positives) > 1:
        multiple_positives += 1
        print(f"Multiple positives found for query: {query}")

    if len(negatives) > 0:
        more_than_one_negative += 1
        print(f"More than one negative found for query: {query}")


print(f"Total queries with multiple positives: {multiple_positives}")
print(f"Total queries with more than one negative: {more_than_one_negative}")

```
