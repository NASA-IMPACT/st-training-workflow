```python
# ...existing code...
import numpy as np
import pandas as pd

def _to_list(x):
    """Convert None / numpy arrays / tuples into a plain Python list."""
    if x is None:
        return []
    if isinstance(x, list):
        return x
    if isinstance(x, tuple):
        return list(x)
    if isinstance(x, np.ndarray):
        return x.tolist()
    # fallback for scalars
    return [x]

def simplify(dx, index=0):
    """
    Flatten a single example (row index) into QA pairs.
    Returns a list of dicts where each dict is one question-answer pair with context/evidence.
    Fields: paper_id (if present), question, question_id, answer_text, evidence (list),
            evidence_text, highlighted_evidence, extractive_spans, unanswerable, yes_no,
            annotation_id, worker_id, search_query, context_fallback
    """
    ex = dx.iloc[index] if isinstance(dx, pd.DataFrame) else dx[index]
    qas = ex.get("qas", {}) if isinstance(ex, dict) else ex["qas"]
    # paper_id = ex.get("id") if isinstance(ex, dict) else ex.get("id", None)

    # derive fallback context from full_text if available
    # full_text = ex.get("full_text", {}) if isinstance(ex, dict) else {}
    # fallback_context = None
    # try:
    #     # paragraphs = full_text.get("paragraphs", []) or []
    #     if paragraphs:
    #         fallback_context = " ".join([" ".join(section) for section in paragraphs if section])
    # except Exception:
    #     fallback_context = None

    qa_pairs = []

    # qas is either dict-of-lists or list-of-dicts
    questions = qas.get("question", [])
    # question_ids = qas.get("question_id", [])
    answers_group = qas.get("answers", [])
    # iterate by question index
    for qi, qtext in enumerate(questions):
        # qid = question_ids[qi] if qi < len(question_ids) else None
        answers_for_q = answers_group[qi] if qi < len(answers_group) else []
        # answers_for_q is a list of annotation-groups
        for ans_group in _to_list(answers_for_q):
            for a in _to_list(ans_group.get("answer", [])):
                evidence_list = _to_list(a.get("evidence"))
                highlighted = _to_list(a.get("highlighted_evidence"))

                unique_evidence = list(set(evidence_list))
                evidence_text = " ".join(map(str, unique_evidence)) if unique_evidence else None

                qa_pairs.append({
                    "question": qtext,
                    "evidence": unique_evidence,
                    "evidence_text": evidence_text,
                    "highlighted_evidence": highlighted,
                })

    return qa_pairs

def flatten_df_to_pairs(dx):
    """Flatten an entire dataframe / dataset split to a pandas DataFrame of QA pairs."""
    all_pairs = []
    for i in range(len(dx)):
        all_pairs.extend(simplify(dx, index=i))
    _dx = pd.DataFrame(all_pairs)

    # drop duplicates
    _dx = _dx.drop_duplicates(subset=["question", "evidence_text"], keep="first")

    return _dx

```

```python
qa_train_df = flatten_df_to_pairs(ds["train"].to_pandas())
qa_val_df = flatten_df_to_pairs(ds["validation"].to_pandas())
qa_test_df = flatten_df_to_pairs(ds["test"].to_pandas())
```
