```python
import pandas as pd

filtered_sde = pd.read_parquet("/rhome/sawale/indus_traning/sentense_transformers/gen_data_stage3/filtered_sde_data/")
original_sde = pd.read_parquet("/rhome/sawale/indus_traning/mlm-fine-tuning/mlm/data/cleaned_prod_dump_filtered_word_count_yake.parquet")
```

```python
# dropping duplicates on both side

# previous shape of dfs
print(f"Filtered SDE shape before dropping duplicates: {filtered_sde.shape}")
print(f"Original SDE shape before dropping duplicates: {original_sde.shape}")

filtered_sde = filtered_sde.drop_duplicates(subset=["id"])
original_sde = original_sde.drop_duplicates(subset=["id"])

# after dropping duplicates
print(f"Filtered SDE shape after dropping duplicates: {filtered_sde.shape}")
print(f"Original SDE shape after dropping duplicates: {original_sde.shape}")

```

```python
# printing the columns of both dataframes
print(f"Filtered SDE columns: {filtered_sde.columns.tolist()}")
print(f"Original SDE columns: {original_sde.columns.tolist()}")
```

```python
# left do left join (filtered_sde, original_sde) based on id, so that i have all the columns
# i want columsn not in filtered_sde but in original_sde to be in the merged table: and dont want any suffixes
# so that i can use the original columns in the filtered_sde

merged_sde = pd.merge(filtered_sde, original_sde, on="id", how="left", suffixes=("", "_original"))
merged_sde.head()
```

```python
# checking the shape of merged_df and of filtered_sde and original_sde
print(f"Merged SDE shape: {merged_sde.shape}")
print(f"Filtered SDE shape: {filtered_sde.shape}")
print(f"Original SDE shape: {original_sde.shape}")
```

```python
# now let's remove all columns of merged_df that has "_original" suffix
merged_sde = merged_sde.loc[:, ~merged_sde.columns.str.endswith("_original")]

merged_sde.head()
```

```python
# lets remove any cmr data points
# shape of df before removing cmr data points
print(f"Merged SDE shape before removing CMR data points: {merged_sde.shape}")
merged_sde = merged_sde[~merged_sde["id"].str.startswith("/SDE/CMR_API/")]
# after removing cmr data points
print(f"Merged SDE shape after removing CMR data points: {merged_sde.shape}")
```

```python
# let remove pds data points
# Shape of df before removing PDS data points
print(f"Merged SDE shape before removing PDS data points: {merged_sde.shape}")

merged_sde = merged_sde[merged_sde["treepath"] != "/Planetary Science/Planetary Data System (PDS)/"]

# after removing PDS data points
print(f"Merged SDE shape after removing PDS data points: {merged_sde.shape}")
```

```python
# get the df that has qa generated
qa_generated_df = pd.read_parquet("/rhome/sawale/indus_traning/sentense_transformers/gen_data_stage3/instructor/data_v2/gpt-4o-mini/")

qa_generated_df.head()
```

```python
qa_generated_df["id"].nunique(), qa_generated_df.shape
```

```python
# now lets remove data points from merged_df that is already in qa_generated_df
print(f"Merged SDE shape before removing QA generated data points: {merged_sde.shape}")
merged_sde = merged_sde[~merged_sde["id"].isin(qa_generated_df["id"].tolist())]
# after removing qa generated data points
print(f"Merged SDE shape after removing QA generated data points: {merged_sde.shape}")
```

```python
# Checing for divisions in collection column

merged_sde["treepath"].str.split("/").str[:2].value_counts()
```

```python
# looks like for BPS, data is very low, we want to get BPS from unfiltered sde data

bps_df = original_sde[original_sde["treepath"].str.startswith("/Biological and Physical Sciences/")]
columns_not_in_bps = set(merged_sde.columns) - set(bps_df.columns)

# adding the columns that are not in bps_df to bps_df.setting value to None
for col in columns_not_in_bps:
    bps_df[col] = None

# before merging order columns in bps_df 
bps_df = bps_df[merged_sde.columns.tolist()]

# also need to remove all bps data points from merged_sde
print(f"Merged SDE shape before removing BPS data points: {merged_sde.shape}")
merged_sde = merged_sde[~merged_sde["treepath"].str.startswith("/Biological and Physical Sciences/")]
# after removing BPS data points
print(f"Merged SDE shape after removing BPS data points: {merged_sde.shape}")

# finally merge the bps_df with merged_sde
merged_sde = pd.concat([merged_sde, bps_df], ignore_index=True)

# checking the shape of merged_sde after merging bps_df
print(f"Merged SDE shape after merging BPS data points: {merged_sde.shape}")
```

```python
merged_sde["treepath"].str.split("/").str[:2].value_counts()
```

```python
# creating new column called division from treepath
merged_sde["division"] = merged_sde["treepath"].str.split("/").str[1:2].str.join("/")

merged_sde["division"].value_counts()
```

```python
# now sampling 2k datapoint from each division: exclude SDE division
# inclause where the sample count is larger than the availabel datapoints, just take all the datapointa
n_sample = 2205
sampled_sde = merged_sde[merged_sde["division"] != "SDE"].groupby("division").apply(lambda x: x.sample(n=min(n_sample, len(x)), random_state=42) if len(x) > n_sample else x).reset_index(drop=True)

sampled_sde["division"].value_counts()

```
