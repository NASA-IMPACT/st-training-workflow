```python
# Function to pretty-print the list for Excel
def pretty_print_list_for_excel(item_list):
    if isinstance(item_list, list):
        # 1. Convert the list to its string representation (e.g., "['apple', 'banana']")
        list_str = str(item_list)

        # 2. Insert newlines for better readability
        # Replace ', ' with ',\n'
        # Handle the opening bracket by adding a newline after it if desired
        # You might also want to remove the outer brackets if they're redundant
        
        # Option A: Simple newline after comma (most common and clean)
        formatted_str = list_str.replace(', ', ',\n ')
        
        # Option B: Add a newline after the opening bracket too (more verbose)
        # formatted_str = list_str.replace('[', '[\n ').replace(', ', ',\n ')
        
        return formatted_str
    return item_list # Return as is if not a list


dx_gpt_4o_mini['questions'] = dx_gpt_4o_mini['questions'].apply(pretty_print_list_for_excel)
dx_gpt_4o_mini['context'] = dx_gpt_4o_mini['context'].apply(pretty_print_list_for_excel)
dx_gpt_4o_mini['search_terms'] = dx_gpt_4o_mini['search_terms'].apply(pretty_print_list_for_excel)
```
