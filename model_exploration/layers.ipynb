```python
from sentence_transformers import (
    SentenceTransformer,
    SentenceTransformerTrainer,
    models,
)
import torch
```

```python
MODEL_NAME = "nasa-impact/indus-sde-st-v0.2"
```

```python
word_embedding_model = models.Transformer(MODEL_NAME)
```

```python
word_embedding_model.get_word_embedding_dimension()
```

```python
local_rank = 0
MODEL_MAX_LEN = 1024
```

```python
model = SentenceTransformer(
            modules=[
                word_embedding_model,
                models.Pooling(
                    word_embedding_model.get_word_embedding_dimension(),
                ),
            ],
            device=f"cuda:{local_rank}",
            tokenizer_kwargs={"model_max_length": MODEL_MAX_LEN, "truncation": True},
            model_kwargs={"torch_dtype": torch.bfloat16},
        )
```

```python
model
```

```python
sentence = "The quick brown fox jumps over the lazy dog."

embedding = model.encode(sentence, convert_to_tensor=True)

print(embedding.size())
embedding[0]
```

```python
from sentence_transformers import SentenceTransformer, models
from torch import  nn

# Step 1: Create the word embedding model
word_embedding_model = models.Transformer(MODEL_NAME)

for p in word_embedding_model.parameters():
    p.requires_grad = False

# Step 2: Add a Dense layer with output dimension 768
dense_model = models.Dense(
    in_features=word_embedding_model.get_word_embedding_dimension(),
    out_features=768,
    activation_function=nn.ReLU()  # You can change this to 'tanh', 'sigmoid', etc.
)

# Step 3: Create the SentenceTransformer model with the new architecture
model = SentenceTransformer(
    modules=[
        word_embedding_model,
        models.Pooling(
            word_embedding_model.get_word_embedding_dimension(),
        ),
        dense_model,
    ],
    device=f"cuda:{local_rank}"
)




sentence = "The quick brown fox jumps over the lazy dog."

embedding = model.encode(sentence, convert_to_tensor=True,)

print(embedding.size())
embedding[0]
```
