# nasa_ads

```python
from datasets import Dataset
import pandas as pd

nasa_ads_data_dir = "/rhome/sawale/indus_traning/sentense_transformers/data/NASA_Embedding_Traindata/"

x0 = pd.read_json(f"{nasa_ads_data_dir}title_abstract_prefix17.jsonl.gz", lines=True, compression='gzip')
x1 = pd.read_json(f"{nasa_ads_data_dir}title_abstract_prefix18.jsonl.gz", lines=True, compression='gzip')
x2 = pd.read_json(f"{nasa_ads_data_dir}title_abstract_prefix19.jsonl.gz", lines=True, compression='gzip')
x3 = pd.read_json(f"{nasa_ads_data_dir}title_abstract_prefix20.jsonl.gz", lines=True, compression='gzip')

df = pd.concat([x0, x1, x2, x3], ignore_index=True)

df["positives"] = df["positives"].apply(lambda x: x["docs"][0])
df = df[["query", "positives"]]

dataset = Dataset.from_pandas(df)

dataset
```

```python
from tqdm import tqdm
# check if htere are multiple positives for a query

multiple_positives = 0
more_than_one_negative = 0

for i in tqdm(range(df.shape[0])):
    positives = df["positives"].iloc[i]["docs"]
    negatives = df["negatives"].iloc[i]["docs"]
    query = df["query"].iloc[i]

    # print(query)
    # print(positives)
    # print(negatives)
    # print("===" * 20)
    
    if len(positives) > 1:
        multiple_positives += 1
        print(f"Multiple positives found for query: {query}")

    if len(negatives) > 0:
        more_than_one_negative += 1
        print(f"More than one negative found for query: {query}")


print(f"Total queries with multiple positives: {multiple_positives}")
print(f"Total queries with more than one negative: {more_than_one_negative}")
```
