```python
from datasets import load_dataset, Features, Value


# Load the streaming dataset
dataset = load_dataset(
    path="rajpurkar/squad_v2",
    streaming=True,
)

# Define the post‐map schema
features = Features({
    "anchor": Value("string"),
    "positive": Value("string"),
})

# Apply the mapping in‐place, keep only the two new fields
dataset = dataset.map(
    lambda ex: {"anchor": ex["text"], "positive": ex["citation"]},
    features=features,
    remove_columns=dataset["train"].column_names,
)

# `dataset` is now an IterableDataset yielding {"anchor":…, "positive":…} pairs

dataset
```
