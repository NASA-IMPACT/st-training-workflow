```python
import numpy as np

sde_search_pairs = sde_data.explode(['search_terms']).reset_index(drop=True)
sde_search_pairs = sde_search_pairs.rename(columns={'search_terms': 'query', 'text': 'context'})

sde_search_pairs["type"] = "search_term-document"
sde_search_pairs["synthesized"] = True


# A helper function to convert numpy types to native Python types
def convert_to_native_python_types(obj):
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    # You can add more checks for other numpy types if needed
    if isinstance(obj, (np.int64, np.int32)):
        return int(obj)
    if isinstance(obj, (np.float64, np.float32)):
        return float(obj)
    # If the object is not a numpy type, return it as is
    return obj


def create_search_metadata(row):
    metadata = {
        "id": row.get('id'),
        "url1": row.get('url1'),
        "title": row.get('title'),
        "text": row.get('text'),
        "prob_included": float(row.get('prob_included')),
        "source": row.get('source'),
        # Apply the conversion to each item in the list
        "questions": [convert_to_native_python_types(q) for q in row.get('questions', [])],
        "_context": [convert_to_native_python_types(c) for c in row.get('_context', [])],
        "request_tokens": int(row.get('request_tokens')),
        "response_tokens": int(row.get('response_tokens')),
        "total_tokens": float(row.get('total_tokens')),
        "time_taken": float(row.get('time_taken'))
    }
    return metadata

columns_to_keep = ["query", "context", "type", "synthesized", "source", "metadata", "url1"]

sde_search_pairs["metadata"] = sde_search_pairs.apply(create_search_metadata, axis=1).apply(json.dumps)
sde_search_pairs = sde_search_pairs[columns_to_keep]
sde_search_pairs.head()

# query	context	type	synthesized	source	metadata	url1
```

```python
import pandas as pd
import json
import os
import numpy as np

def sanitize_metadata(metadata):
    """
    Recursively sanitizes a metadata dictionary to ensure all primitive
    values are strings, preventing type errors during dataset loading.

    Args:
        metadata: The dictionary or list to sanitize.

    Returns:
        The sanitized dictionary or list.
    """
    if isinstance(metadata, dict):
        return {str(k): sanitize_metadata(v) for k, v in metadata.items()}
    elif isinstance(metadata, list):
        return [sanitize_metadata(elem) for elem in metadata]
    elif metadata is None:
        return None
    # Convert all other primitive types (int, float, bool) to string
    elif not isinstance(metadata, str):
        return str(metadata)
    return metadata


def create_sentence_transformer_dataset(
    df: pd.DataFrame,
    output_dir: str = "huggingface_dataset",
    train_split: float = 0.90,
    val_split: float = 0.05,
    random_state: int = 42
):
    """
    Shuffles, splits, and converts a DataFrame into the JSONL and TSV format
    for sentence-transformer models, while sanitizing metadata to prevent
    type errors during loading.

    Args:
        df (pd.DataFrame): Input DataFrame with columns 'query', 'context',
                           'query_metadata', 'corpus_metadata'.
        output_dir (str): Directory to save the dataset files.
        train_split (float): Proportion for the training set.
        val_split (float): Proportion for the validation set.
        random_state (int): Seed for reproducible shuffling.
    """
    # --- 0. Validate Splits ---
    test_split = 1.0 - train_split - val_split
    if not np.isclose(train_split + val_split + test_split, 1.0):
        raise ValueError("Train, validation, and test splits must sum to 1.0")
    if train_split <= 0 or val_split < 0 or test_split < 0:
        raise ValueError("Train split must be positive, and val/test splits must be non-negative.")

    # --- 1. Shuffle and Split DataFrame ---
    df_shuffled = df.sample(frac=1, random_state=random_state)
    train_end = int(train_split * len(df_shuffled))
    val_end = int((train_split + val_split) * len(df_shuffled))
    train_df, val_df, test_df = df_shuffled.iloc[:train_end], df_shuffled.iloc[train_end:val_end], df_shuffled.iloc[val_end:]
    print(f"Dataset split: {len(train_df)} train, {len(val_df)} validation, {len(test_df)} test examples.")

    # --- 2. Create Output Directories ---
    qrels_dir = os.path.join(output_dir, "qrels")
    os.makedirs(qrels_dir, exist_ok=True)
    print(f"Created directory structure at: {output_dir}")

    # --- 3. Generate Master Corpus and Queries Files ---
    corpus_dict = {}
    corpus_id_counter = 0
    print("Generating master corpus.jsonl and queries.jsonl files with sanitized metadata...")

    with open(os.path.join(output_dir, "corpus.jsonl"), "w", encoding="utf-8") as corpus_file, \
         open(os.path.join(output_dir, "queries.jsonl"), "w", encoding="utf-8") as queries_file:

        for i, row in df_shuffled.iterrows():
            query_id = f"q{i}"
            context_text = row['context']

            # Sanitize metadata before using it
            sanitized_corpus_meta = sanitize_metadata(row['corpus_metadata'])
            sanitized_query_meta = sanitize_metadata(row['query_metadata'])

            # Process and write corpus entry (if unique)
            if context_text not in corpus_dict:
                corpus_id_counter += 1
                corpus_id = f"c{corpus_id_counter}"
                corpus_dict[context_text] = corpus_id
                corpus_entry = {
                    "_id": corpus_id, "text": context_text, "title": "",
                    "metadata": sanitized_corpus_meta
                }
                corpus_file.write(json.dumps(corpus_entry) + "\n")

            # Process and write query entry
            query_entry = {
                "_id": query_id, "text": row['query'],
                "metadata": sanitized_query_meta
            }
            queries_file.write(json.dumps(query_entry) + "\n")

    # --- 4. Generate Qrels Files for Each Split ---
    def write_qrels(split_df, split_name):
        print(f"Generating qrels/{split_name}.tsv...")
        with open(os.path.join(qrels_dir, f"{split_name}.tsv"), "w", encoding="utf-8") as qrels_file:
            qrels_file.write("query-id\tcorpus-id\tscore\n")
            for i, row in split_df.iterrows():
                query_id = f"q{i}"
                corpus_id = corpus_dict[row['context']]
                qrels_file.write(f"{query_id}\t{corpus_id}\t1\n")

    write_qrels(train_df, "train")
    if not val_df.empty: write_qrels(val_df, "val")
    if not test_df.empty: write_qrels(test_df, "test")

    print("\nSuccessfully finished generating dataset files.")

```

```python
import wandb
import pandas as pd

# --- Configuration: SET THESE VALUES ---
# 1. Path of the run you want to copy (e.g., "team-a/source-project/abc123xyz")
SOURCE_RUN_PATH = "nasa-impact/nasa_st_traning/9olox5mz" 

# 2. Name of the new destination project
DESTINATION_PROJECT = "nasa-indus-sde-s2"

# 3. Name of the destination entity (your username or team name)
DESTINATION_ENTITY = "impact-ibm-collaboration"

# 4. (Optional) Name for the new run. If None, W&B will generate one.
NEW_RUN_NAME = "stage1" 
# --- End of Configuration ---


def copy_run(source_run_path, dest_entity, dest_project, new_run_name=None):
    """
    Copies a W&B run to a new project.

    This function fetches the config, summary, metrics history, and files
    from a source run and logs them to a new run in the destination.
    
    Args:
        source_run_path (str): The path of the source run (entity/project/run_id).
        dest_entity (str): The entity (username or team) for the new run.
        dest_project (str): The project name for the new run.
        new_run_name (str, optional): A specific name for the new run. Defaults to None.
    """
    try:
        # Initialize the W&B API
        api = wandb.Api()
        
        print(f"Fetching source run: {source_run_path}")
        source_run = api.run(source_run_path)

        # --- 1. Fetch Configuration and Summary ---
        source_config = source_run.config
        source_summary = source_run.summary._json_dict

        # --- 2. Initialize the New Run ---
        new_run = wandb.init(
            project=dest_project,
            entity=dest_entity,
            config=source_config,
            name=new_run_name,
            reinit=True # Use reinit=True if running in a notebook
        )
        print(f"Created new run: {new_run.path}")

        # --- 3. Fetch and Log Metrics History ---
        print("Fetching and logging metrics history...")
        history_df = source_run.history()
        
        # Stream the history back to the new run
        for index, row in history_df.iterrows():
            # Convert row to a dictionary and filter out NaNs and the _step column
            log_dict = {k: v for k, v in row.to_dict().items() if pd.notna(v) and k != '_step'}
            wandb.log(log_dict, step=int(row['_step']))

        # --- 4. Fetch, Download, and Log Files/Artifacts ---
        print("Copying saved files and artifacts...")
        for f in source_run.files():
            # Download the file to the local directory
            f.download(replace=True)
            
            # Log files saved with wandb.save() to the new run
            if f.name.startswith('code/') or f.name in ['config.yaml', 'output.log', 'requirements.txt']:
                 # These are system files, log them carefully or skip
                 pass # Often better to let the new run generate its own system files
            else:
                wandb.save(f.name)

        # For artifacts (versioned data/models)
        for artifact in source_run.logged_artifacts():
            # To simply link the artifact to the new run without re-uploading:
            new_run.use_artifact(artifact)
            # If you want to create a new, distinct version of the artifact in the new project:
            # (This is less common for a simple copy)
            # downloaded_artifact = new_run.use_artifact(artifact.name)
            # new_artifact = wandb.Artifact(name=downloaded_artifact.name.split(':')[0], type=downloaded_artifact.type)
            # new_artifact.add_dir(downloaded_artifact.download())
            # new_run.log_artifact(new_artifact)

        # --- 5. Update the Final Summary ---
        wandb.summary.update(source_summary)

        # --- 6. Finish the New Run ---
        new_run.finish()
        print(f"\n✅ Successfully copied run to: {new_run.url}")

    except Exception as e:
        print(f"An error occurred: {e}")
        if 'new_run' in locals() and new_run is not None:
            new_run.finish(exit_code=1) # Mark the new run as crashed


# --- Run the copy process ---
if __name__ == "__main__":
    copy_run(
        source_run_path=SOURCE_RUN_PATH,
        dest_entity=DESTINATION_ENTITY,
        dest_project=DESTINATION_PROJECT,
        new_run_name=NEW_RUN_NAME
    )
```
